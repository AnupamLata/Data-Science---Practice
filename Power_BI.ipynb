# **What  is Power BI ?**

---
* Power BI is a Business Intelligence & Data Visualization tool that converts data from various data source into interactive dashboards and analysis reports.


* Power BI takes raw data ➡ analyzes it ➡ turns it into meaningfuk charts , graphs, reports, and dashboards.





***For example,***

suppose a company has an Excel file containing 10,000 sales records. Instead of manually analyzing them, Power BI can create a dashboard showing:
* Total sales
* Monthly Sales Trend
* Best-selling Products
* Top Customers
* Sales by City/State
* Profit and Loss

# **How Power BI works:**

---
Data ➡ Clean ➡ Transform ➡ Analyze ➡ Visualize ➡ Dashboard

* Get Data : Power BI can connect to Excel , CSV, SQL , Server, databases, websites, etc.
* Clean & Transform Data : You can remove duplicates, handle missing values, change data types , merge tables ,etc.

  this is mainly done using Power Query.

* Create Data Model: You can connect different tables using relationships.  

  Customers ↔ Orders ↔ Products

* Analyze Data : Power BI uses DAX(Data Analysis Expressions) to create calculations such as :
  * Total Sales
  * Average Sales
  * Profits %
  * Year-over-year Growth

* Create Visualizations: you can make :  
     * Bar charts
     * Line charts
     * pie / donut charts
     * Tables
     * Cards
     * maps
     * Slicers

* Create Dashboard / Report : All these visuals can be arranged into an interactive report.     


**Power BI** is a Microsoft tool that converts raw data into interactive , understandable reports and dashboards so people can make better decisions.
